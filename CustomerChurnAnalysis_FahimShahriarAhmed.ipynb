# Data Visualization


```python
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
# Box-plot of Monthly Charges by Churn

sb.set_style("darkgrid")
sb.boxplot(x= 'Churn',
          y= 'MonthlyCharges',
          data= df)


# Jaader monthly charges beshi tara churn hoitese
```

```python
# Box-plot of Monthly Charges by Churn and divided by Contract

sb.set_style("darkgrid")
plt.figure(figsize=(10,5))
sb.boxplot(x= 'Churn',
          y= 'MonthlyCharges',
          hue = 'Contract',
          data= df)
plt.legend(title='Contract', loc='upper left')

#long term contract+behsi charge er manush beshi churn hoitese
```

```python
# Box-plot of Total Charges by Churn and divided by Contract

plt.figure(figsize=(10,5))
sb.boxplot(x= 'Churn',
          y= 'TotalCharges',
          hue = 'Contract',
          data= df)
plt.legend(loc= 'upper left')

# jaader charges beshi tara churn hoitese
```

```python
# Box-plot of Monthly Charges by Churn divided by gender

plt.figure(figsize=(10,5))
sb.boxplot(x= 'Churn',
          y= 'MonthlyCharges',
          hue = 'gender',
          data= df)
plt.legend(loc= 'upper right')

# gennder does not change anything for churn
```

```python
# Box-plot of Total Charges by Churn and divided by payment method

plt.figure(figsize=(10,5))
sb.boxplot(x= 'Churn',
          y= 'TotalCharges',
          hue = 'PaymentMethod',
          data= df)
plt.legend(loc= 'upper right')
```

```python
plt.figure(figsize=(12, 5))
sns.histplot(data=df, x='MonthlyCharges', hue='Churn', kde=True)
plt.title('Observing the Distribution of MonthlyCharges based on the Churn column')
plt.show()

#monthly charges jaader kom tader churn beshi(subscription off kore dicche)
```

```python
plt.figure(figsize=(12, 5))
sns.histplot(data=df, x='tenure', hue='Churn', kde=True)
plt.title('Observing the Distribution of MonthlyCharges based on the Churn column')
plt.show()

# tenure jaader korm tader churn rate beshi
```

```python
plt.figure(figsize=(5,5))
sns.scatterplot(x='tenure', y='MonthlyCharges', hue='Churn', data=df)
plt.title('Observing the Balance based on the Churn column')
plt.show()

#monthly charges beshi but tenure kom emon customer er churn rate beshi
```

# Business Recommendation

```python
#Customer who has lower tenor and high monthly payments are more prone to churn.
# We have to provide them value for value-for-money products to give them reasons to stay.
# We have found that those customer who has more months of stay with us are likely to stay as customers.
# We have seen that our additional product like online security and online backup are used less by our customers.
# We can give them a free 6-month trial to use them so that customers can find our product as value for money.
# Thus increasing their stay with us and therefore, eventually lowering the churn rate.
```
